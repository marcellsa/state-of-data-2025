### Hipóteses de Pesquisa

#### Hipóteses a serem exploradas

- **H1**: Profissionais 40+ enfrentam mais dificuldade para alcançar oportunidades.
- **H2**: Entre profissionais de dados que já atuam na área, pessoas 40+ tendem a ocupar faixas salariais inferiores às de profissionais mais jovens com o mesmo nível de atuação.
- **H3**: Entre os respondentes da pesquisa, profissionais 40+ tendem a ser menos aproveitados tanto em funções técnicas quanto de liderança, ocupando posições que nem sempre refletem seu perfil ou objetivo de carreira.

##### Hipótese 1 - Profissionais 40+ enfrentam mais dificuldade para alcançar oportunidades

**Pergunta**: entre os entrevistados que atuam ou buscam ativamente atuação em dados, profissionais 40+ demonstram maior dificuldade em alcançar oportunidades de trabalho do que os mais jovens.

**Objetivo da análise**: investigar se há indícios de desigualdade de acesso a oportunidades com base na faixa etária, usando como indicadores:

- Tipo de oportunidade desejada (`5.b_oportunidade_buscada`)
- Tempo de busca por vaga (`5.c_tempo_em_busca_de_oportunidade`)
- Experiência em processos seletivos (`5.d_experiencia_em_processos_seletivos`)


```python
# Filtrando apenas entrevistados que informaram tempo de busca por vaga
df_h1 = df[df["5.c_tempo_em_busca_de_oportunidade"].notna()].copy()

# Criando uma coluna simplificada para grupos etários: 40+ vs <40
df_h1["grupo_idade"] = df_h1["1.a.1_faixa_idade"].apply(
    lambda x: (
        "40+"
        if x
        in [
            "de 40 a 44 anos",
            "de 45 a 49 anos",
            "de 50 a 54 anos",
            "de 55 a 59 anos",
            "60 anos ou mais",
        ]
        else "<40"
    )
)
```

```python
# 1. Tempo em busca de oportunidade, grupo 40+
# Ordenando as categorias
ordem_tempo = [
    "0 - 6 meses",
    "7 meses - 1 ano",
    "1 ano - 2 anos",
    "acima de 2 anos"
]

g6 = px.histogram(
    df_h1,
    x="5.c_tempo_em_busca_de_oportunidade",
    category_orders={"5.c_tempo_em_busca_de_oportunidade": ordem_tempo},
    text_auto=True,
    title="Tempo em busca de oportunidade em Dados (Grupo 40+)",
    labels={"5.c_tempo_em_busca_de_oportunidade": "Tempo de Busca"},
)
g6.update_layout(xaxis_title="Tempo em busca de oportunidade", yaxis_title="Nº de Pessoas")
g6.show()
```

```python
# 2. Participação em Processos Seletivos
# Contando e ordenando as categorias da mais frequente para a menos
ordem_experiencia = (
    df_h1["5.d_experiencia_em_processos_seletivos"]
    .value_counts()
    .sort_values(ascending=False)
    .index.tolist()
)


g7 = px.histogram(
    df_h1,
    y="5.d_experiencia_em_processos_seletivos",
    orientation="h",
    text_auto=True,
    title="Participação em Processos Seletivos (Grupo 40+)",
    category_orders={"5.d_experiencia_em_processos_seletivos": ordem_experiencia},
    labels={
        "5.d_experiencia_em_processos_seletivos": "Experiência em Processos Seletivos"
    },
)
g7.update_layout(
    xaxis_title="Nº de Pessoas", yaxis_title="Experiência em Processos Seletivos",
)
g7.show()
```

```python
# 3. Tipo de oportunidade buscada
# Contandp e ordenando as categorias da mais para a menos buscada
ordem_oportunidade = (
    df_h1["5.b_oportunidade_buscada"]
    .value_counts()
    .sort_values(ascending=False)
    .index.tolist()
)

g8 = px.histogram(
    df_h1,
    y="5.b_oportunidade_buscada",
    orientation="h",
    text_auto=True,
    title="Tipo de Oportunidade Buscada (Grupo 40+)",
    category_orders={"5.b_oportunidade_buscada": ordem_oportunidade},
    labels={"5.b_oportunidade_buscada": "Oportunidade Buscada"},
)

g8.update_layout(xaxis_title="Nº de Pessoas", yaxis_title="Oportunidade Buscada")

g8.show()
```

##### Conclusão — Hipótese 1

A hipótese investigava se, entre profissionais de dados, pessoas com 40 anos ou mais enfrentam mais dificuldades para conquistar oportunidades do que profissionais mais jovens.

**Principais achados:**

- **Tempo de busca por vagas:** O grupo 40+ apresenta uma distribuição semelhante ao grupo <40, com destaque para busca entre 3 a 6 meses em ambos os grupos.
  
- **Participação em processos seletivos:** A maioria dos respondentes (de todas as idades) participou de poucos ou nenhum processo seletivo, mas o grupo 40+ mostra ligeira tendência a participar menos.

- **Tipo de oportunidade buscada:** Pessoas 40+ buscam majoritariamente oportunidades como analistas e especialistas, e poucas estão focadas em posições de gestão, o que indica desejo de inserção técnica no setor.

**Interpretação:**

Embora os dados não apresentem diferenças gritantes entre os grupos etários, há sinais de que:
- O grupo 40+ **participa menos de processos seletivos**;
- Está em busca ativa de recolocação, porém **não necessariamente direcionado a cargos de liderança**;
- Pode estar enfrentando **barreiras mais sutis**, como **filtro etário invisível em processos seletivos**, o que não aparece diretamente nos dados, mas pode ser complementado em análises qualitativas futuras.
