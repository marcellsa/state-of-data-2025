### Análise Exploratória de Dados - State of Data 2024-2025

#### Contextualização e Objetivos

O mercado de dados está em constante crescimento, mas será que ele é acessível para profissionais em todas as fases da vida? Em áreas que mudam rápido, como TI e marketing digital, pessoas com mais de 40 anos podem enfrentar obstáculos como preconceito por idade e defasagem técnica.

Com base na pesquisa State of Data Brazil 2024-2025, este estudo investiga os principais desafios vividos por profissionais 40+ no setor de dados.

**Objetivos:**
1. Investigar se há desigualdade de oportunidades e crescimento para profissionais 40+.
2. Avaliar a percepção de discriminação por idade no mercado.
3. Entender se há defasagem técnica percebida ou real entre profissionais mais velhos.

### Análise das Hipóteses

#### A partir da exploração inicial dos dados, três hipóteses principais foram formuladas sobre os desafios enfrentados por profissionais com 40 anos ou mais no mercado de dados. A seguir, analisamos cada uma com base nas variáveis da pesquisa.

#### H1. Profissionais 40+ enfrentam mais dificuldade para progredir na carreira?

Objetivo: Verificar se os profissionais com 40 anos ou mais estão concentrados em níveis mais baixos de senioridade.

- Variáveis:
  - `grupo_idade` (derivada de `1.a.1_faixa_idade`)
  - `2.g_nivel` (nível de senioridade)

```python
g_h1 = px.histogram(
    df,
    x="2.g_nivel",
    color="grupo_idade",
    barmode="group",
    title="Distribuição de Nível de Senioridade por Grupo Etário",
    labels={"2.g_nivel": "Nível de Senioridade", "grupo_idade": "Grupo Etário"},
)

g_h1.update_layout(xaxis_title="Senioridade", yaxis_title="Nº de Pessoas")

g_h1.show()
```

**Interpretação:**

O gráfico mostra que o grupo 40+ está mais representado em cargos Sênior e de Gestão do que o grupo mais jovem — o que pode ser esperado. No entanto, ainda há presença significativa de profissionais 40+ em níveis Júnior e Pleno.

Isso pode indicar dois cenários:

- Parte desses profissionais pode estar em transição de carreira, começando na área de dados com mais idade.
- Ou ainda há barreiras para progressão, mesmo com experiência acumulada.

Esse padrão reforça a necessidade de explorar outras dimensões, como tempo de experiência, percepção de discriminação e defasagem técnica.

#### H2. Profissionais 40+ estão mais concentrados em áreas menos técnicas ou mais voltadas à gestão?
Objetivo: Investigar se profissionais com 40 anos ou mais ocupam mais cargos de gestão ou funções menos técnicas, enquanto profissionais mais jovens estão majoritariamente em cargos técnicos.

Variáveis sugeridas:

- grupo_idade (derivada de 1.a.1_faixa_idade)

- 2.d_atua_como_gestor (indica se exerce cargo de gestão)

- 2.f_cargo_atual ou 4.a_funcao_de_atuacao (para classificar áreas técnicas vs. não técnicas)

```python
# Verificar os valores únicos da coluna de cargo
df["2.f_cargo_atual"].value_counts()
```

```python
# Função refinada para classificar os cargos
def classificar_cargo(cargo):
    if pd.isnull(cargo):
        return "Não informado"

    cargo = cargo.lower()

    # Gestão
    if (
        "product manager" in cargo
        or "pm" in cargo
        or "dpm" in cargo
        or "gpm" in cargo
        or "po" in cargo
    ):
        return "Gestão"

    # Técnico
    elif any(
        palavra in cargo
        for palavra in [
            "analista de dados",
            "data analyst",
            "cientista de dados",
            "data scientist",
            "engenheiro de dados",
            "data engineer",
            "analytics engineer",
            "bi analyst",
            "analista de bi",
            "analista de negócios",
            "business analyst",
            "engenheiro de software",
            "desenvolvedor",
            "ml engineer",
            "engenheiro de machine learning",
            "ai engineer",
            "arquiteto de dados",
            "estatístico",
            "analista técnico",
            "analista de suporte",
        ]
    ):
        return "Técnico"

    # Outros
    else:
        return "Outros"
```

```python
# Aplicando a função no dataframe
df["categoria_cargo"] = df["2.f_cargo_atual"].apply(classificar_cargo)
```

```python
# Criar tabela com proporções
tabela_cargo = (
    df.groupby(["grupo_idade", "categoria_cargo"]).size().reset_index(name="quantidade")
)

tabela_cargo["proporcao"] = tabela_cargo.groupby("grupo_idade")["quantidade"].transform(
    lambda x: x / x.sum()
)
```

```python
g_h2 = px.bar(
    tabela_cargo,
    x="grupo_idade",
    y="proporcao",
    color="categoria_cargo",
    barmode="group",
    text="quantidade",
    title="Distribuição de Tipos de Cargo por Grupo Etário",
    labels={
        "grupo_idade": "Grupo Etário",
        "proporcao": "Proporção",
        "categoria_cargo": "Tipo de Cargo",
    },
)
g_h2.update_layout(yaxis_tickformat=".0%", xaxis_title=None)
g_h2.show()
```

A hipótese 2 é parcialmente confirmada:

Profissionais mais jovens (<40) estão mais concentrados em cargos técnicos.

Profissionais 40+ têm menos presença técnica proporcionalmente, mas não necessariamente mais presença em cargos de gestão.

A alta taxa de respostas “Não informado” no grupo 40+ sugere que a análise pode estar subestimando ou distorcendo a realidade desse grupo.

#### H3. Profissionais 40+ apresentam menor domínio ou uso de ferramentas modernas?

Objetivo: Verificar se profissionais 40+ usam menos ferramentas populares ou modernas de dados.

- Variáveis:
  - `grupo_idade`
  - Perguntas multivaloradas de ferramentas (ex: `4.b_ferramentas_de_visualizacao_` ou `7.c_ferramentas_de_analise_`)

Sugestão de gráfico:
- Heatmap ou barras empilhadas: **Uso de ferramentas por grupo etário**
- Pode-se agrupar ferramentas por categorias (ex: modernas vs tradicionais)

```python
# Seleção de colunas relevantes
colunas_ferramentas = [
    # Linguagens
    "4.d.1_SQL",
    "4.d.3_Python",
    # BI
    "4.j.1_Microsoft PowerBI",
    "4.j.3_Tableau",
    "4.j.7_Looker",
    # Cloud
    "4.h.1_Amazon Web Services (AWS)",
    "4.h.2_Google Cloud (GCP)",
    "4.h.3_Azure (Microsoft)",
    # Plataformas modernas
    "4.g.22_Google BigQuery",
    "4.g.26_Snowflake",
    "4.g.27_Databricks",
]
```

```python
# Copiar dados apenas das colunas necessárias
df_tools = df[["grupo_idade"] + colunas_ferramentas].copy()

# Converter para formato longo
df_melt = df_tools.melt(id_vars="grupo_idade", var_name="Ferramenta", value_name="Usa")

# Filtrar respostas positivas (uso declarado)
df_melt = df_melt[df_melt["Usa"] == True]

# Agrupar por grupo etário e ferramenta
uso_por_idade = (
    df_melt.groupby(["grupo_idade", "Ferramenta"]).size().reset_index(name="Quantidade")
)

# Total por grupo para calcular proporção
total_por_grupo = df.groupby("grupo_idade").size().to_dict()
uso_por_idade["Proporcao"] = uso_por_idade.apply(
    lambda row: row["Quantidade"] / total_por_grupo.get(row["grupo_idade"], 1), axis=1
)
```

```python
# Gráfico
fig = px.bar(
    uso_por_idade,
    x="Ferramenta",
    y="Proporcao",
    color="grupo_idade",
    barmode="group",
    title="Adoção de Ferramentas por Grupo Etário",
    labels={"Proporcao": "Proporção de Uso", "grupo_idade": "Faixa Etária"},
    text="Quantidade",
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

```python
# Outra versão
# Filtrar colunas relevantes (exemplo com visualização + análise + BI)
cols_visual = [col for col in df.columns if col.startswith("4.b.")]
cols_analise = [col for col in df.columns if col.startswith("7.c.")]
cols_bi = [col for col in df.columns if col.startswith("4.j.")]
cols_total = cols_visual + cols_analise + cols_bi

# Copiar e transformar para formato longo
df_melt = df[["grupo_idade"] + cols_total].melt(
    id_vars="grupo_idade", var_name="ferramenta", value_name="usa"
)
df_melt = df_melt[df_melt["usa"] == True]

# Agrupar por grupo etário e ferramenta
uso = (
    df_melt.groupby(["grupo_idade", "ferramenta"]).size().reset_index(name="quantidade")
)

# Total por grupo
total_grupo = df["grupo_idade"].value_counts().to_dict()
uso["proporcao"] = uso.apply(
    lambda row: row["quantidade"] / total_grupo[row["grupo_idade"]], axis=1
)

# Dicionário de categorização das ferramentas
mapa_modernas = {
    "powerbi": "Tradicional",
    "excel": "Tradicional",
    "tableau": "Tradicional",
    "spss": "Tradicional",
    "sas": "Tradicional",
    "looker": "Moderna",
    "snowflake": "Moderna",
    "databricks": "Moderna",
    "bigquery": "Moderna",
    "dbt": "Moderna",
    "chatgpt": "Moderna",
}


# Criar nova coluna com categoria (moderna vs tradicional)
def classificar_categoria(ferramenta):
    f = ferramenta.lower()
    for chave in mapa_modernas:
        if chave in f:
            return mapa_modernas[chave]
    return "Outros/Não categorizado"


uso["categoria"] = uso["ferramenta"].apply(classificar_categoria)

# Agrupar por grupo etário e categoria
uso_categoria = (
    uso.groupby(["grupo_idade", "categoria"])["quantidade"].sum().reset_index()
)

uso_categoria["proporcao"] = uso_categoria.apply(
    lambda row: row["quantidade"] / total_grupo[row["grupo_idade"]], axis=1
)
```

```python
fig = px.bar(
    uso_categoria,
    x="grupo_idade",
    y="proporcao",
    color="categoria",
    text="quantidade",
    barmode="stack",
    title="Proporção de Uso de Ferramentas por Categoria e Faixa Etária",
    labels={
        "grupo_idade": "Grupo Etário",
        "proporcao": "Proporção",
        "categoria": "Categoria de Ferramenta",
    },
)
fig.update_layout(yaxis_tickformat=".0%")
fig.show()
```
